```python
import os
import pandas as pd
```

Post-hoc curate the metadata CSV file for the `3_maetest` subset, sourcing and referencing the `tri24` meta file for the `maestro_subset` originating in the `mpteval` dataset repo: https://zenodo.org/records/13897703

```python
df = pd.read_csv('../tri24/data/meta_csv/meta_maestro_subset.csv')

new_records = []
for i, record in enumerate(df[df['split'] == 'test'].to_dict('records')):
    path_midi = record['midi_performance']
    path_midi = f'../tri24/data/maestro_subset/test/{path_midi}'
    midi_base = os.path.basename(path_midi)
    if midi_base == 'KOLESO06M.mid':
        identifier = record['folder'].split('/')[-1]
        midi_base = midi_base.replace('.mid', f'_{identifier}.mid')
    
    new_record = {**{'mds_file': midi_base.replace('.mid', '')}, **record}
    new_records.append(new_record)

# save new metadata as a csv
df_new = pd.DataFrame(new_records)
df_new = df_new[['mds_file', 'title', 'composer', 'midi_performance', 'midi_path', 'duration_sec']]

# rename columns
df_new.columns = ['mds_file', 'title', 'composer', 'path_in_mpteval', 'path_in_maestro', 'duration_sec']
# df_new.to_csv('../data/metadata/3_maetest_meta.csv', index=False)
df_new
```
